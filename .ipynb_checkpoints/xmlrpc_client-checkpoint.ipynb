# ЛР3. Клиент

Тетрадь подготовлена на основе предыдущей ЛР и требований ЛР3.

```python
import xmlrpc.client
import pickle
import numpy as np
from PIL import Image
from pylab import *
%matplotlib inline

server=xmlrpc.client.ServerProxy("http://localhost:8010",allow_none=True)
stats=xmlrpc.client.ServerProxy("http://localhost:8009",allow_none=True)
print("Методы прокси:", server.system.listMethods())
print("Ping:",server.ping()); print("Now:",server.now()); print("Type:",server.type(2)); print("Sum:",server.sum(2,3)); print("Pow:",server.pow(2,3))

print("\nПроверка чёрного списка")
for args in [('Иванов','Иван','Иванович','22.03.1989'),('иванов','иван','иванович','22.03.1989'),('Иваноа','Иван','Иванович','22.03.1989'),('Сидоров','Пётр','Петрович','01.01.1990')]: print(args,"=>",server.black_list_check(*args))

img_color=Image.open('Jellyfish.jpg'); img_arr_color=np.array(img_color)
img_mono=Image.open('11.bmp'); img_arr_mono=np.array(img_mono)
pimg_color=xmlrpc.client.Binary(pickle.dumps(img_arr_color)); pimg_mono=xmlrpc.client.Binary(pickle.dumps(img_arr_mono))
threshold=40
inv_color=pickle.loads(server.color_inversion(pimg_color).data); inv_mono=pickle.loads(server.color_inversion(pimg_mono).data)
bw_color=pickle.loads(server.binarize_image(pimg_color,threshold).data); bw_mono=pickle.loads(server.binarize_image(pimg_mono,threshold).data)
flip_color=pickle.loads(server.flip_vertical(pimg_color).data); flip_mono=pickle.loads(server.flip_vertical(pimg_mono).data)
print(f"Облачность color: {server.cloudiness_percent(pimg_color,threshold):.2f}%")
print(f"Облачность mono: {server.cloudiness_percent(pimg_mono,threshold):.2f}%")

print("\nПроверка Rate limiting: 6 вызовов ping подряд")
for i in range(6): print(i+1,server.ping())

print("\nЖурнал событий")
for e in stats.get_events('','','','','')[-20:]: print(e)
print("\nСрез sum:",stats.get_events('sum','','','',''))
print("Срез по времени:",stats.get_events('','2020-01-01 00:00:00','2030-12-31 23:59:59.999999','','')[:10])
print("Срез duration >= 0:",stats.get_events('','','','0','')[:10])
print("Общая статистика:",stats.get_stats())

print("\nОтказоустойчивость: остановите stats-сервер и повторите server.sum(2,3).")
print("Результат должен остаться 5, так как прокси не делает статистику обязательной.")
```
